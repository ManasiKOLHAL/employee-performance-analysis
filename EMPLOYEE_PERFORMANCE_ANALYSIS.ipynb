### Chi-Square Test

````python
from scipy.stats import chi2_contingency

categorical_features = [
    "Gender",
    "EducationBackground",
    "MaritalStatus",
    "EmpDepartment",
    "EmpJobRole",
    "BusinessTravelFrequency",
    "OverTime",
    "Attrition"
]

chi_square_results = []

for feature in categorical_features:
    contingency_table = pd.crosstab(
        df_working[feature],
        df_working["PerformanceRating"]
    )

    chi2, p, dof, expected = chi2_contingency(contingency_table)

    chi_square_results.append({
        "Feature": feature,
        "Chi2 Statistic": round(chi2, 2),
        "P-Value": round(p, 4)
    })

chi_square_df = pd.DataFrame(chi_square_results)

chi_square_df
````

- p < 0.05: Significant association with performance.
- p ≥ 0.05: No statistically significant association.

#### Conclusion

- The Chi-Square test identifies **EmpDepartment**, **EmpJobRole**, and **OverTime** as the only categorical variables that have a statistically significant relationship with **Performance Rating** (**p < 0.05**).
- The remaining categorical features do not show a significant association with employee performance and are less influential in explaining performance differences.
```

### One-Way ANOVA

ANOVA determines whether the mean age differs significantly across performance rating groups. The same approach can be applied to other numerical variables if needed.

- This indicates that **Age is not a significant factor influencing employee performance** in the dataset.
- Employees with different performance ratings have **similar average ages**, suggesting performance is independent of age.
- Other factors such as **Environment Satisfaction**, **Last Salary Hike Percentage**, and **Work-Life Balance** are likely to have a greater impact on performance than age.

### Conclusion

- The One-Way ANOVA test confirms that **Age does not significantly differ across Performance Rating groups (p > 0.05)**.
- Hence, **Age is not considered a key predictor of employee performance** in this analysis.

Feature Engineering is the process of creating new meaningful variables from existing data to improve model performance and business understanding.

Feature Selection identifies the most relevant variables while removing redundant or non-informative features. This reduces model complexity, improves generalization, and enhances interpretability.

### Create Career Growth Ratio

Employees who have remained in the company for many years without promotion may experience slower career progression.

````python
feature_df["CareerGrowthRatio"] = (
    feature_df["ExperienceYearsAtThisCompany"] /
    (feature_df["YearsSinceLastPromotion"] + 1)
)
````

A higher ratio indicates more recent promotions relative to tenure, suggesting healthier career progression.

### Crate Average Experience Per Company

````python
feature_df["AverageExperiencePerCompany"] = (
    feature_df["TotalWorkExperienceInYears"] /
    (feature_df["NumCompaniesWorked"] + 1)
)
````

This feature estimates job stability. Employees with longer average tenure at previous companies may exhibit greater organizational commitment.

### Create Satisfaction Index 

````python
feature_df["SatisfactionIndex"] = (
    feature_df[
        [
            "EmpEnvironmentSatisfaction",
            "EmpJobSatisfaction",
            "EmpRelationshipSatisfaction",
            "EmpWorkLifeBalance"
        ]
    ].mean(axis=1)
)
````

This engineered feature captures an employee's overall workplace satisfaction, which is expected to be a strong predictor of performance.

### Create Experience Ratio

````python
feature_df["CompanyExperienceRatio"] = (
    feature_df["ExperienceYearsAtThisCompany"] /
    feature_df["TotalWorkExperienceInYears"].replace(0, 1)
)
````

A higher ratio suggests employees have spent a greater proportion of their careers at INX Future Inc., potentially reflecting loyalty and organizational familiarity.

### Create Promotion Delay Indicator

````python
feature_df["PromotionDelay"] = np.where(
    feature_df["YearsSinceLastPromotion"] >= 5,
    1,
    0
)
````

Employees with extended periods without promotion may require career development discussions or advancement opportunities.

### Create Overtime Indicator

````python
feature_df["OverTimeFlag"] = feature_df["OverTime"].map({
    "Yes": 1,
    "No": 0
})
````

Before training machine learning models, the dataset must be transformed into a suitable numerical format.

### Define Feature Matrix and Target Variable

````python
X = encoded_df.drop("PerformanceRating", axis=1)

y = encoded_df["PerformanceRating"]
````

### Train-Test Split

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)

print("Training Samples :", X_train.shape)

print("Testing Samples :", X_test.shape)
````

Here using stratified sampling preserves the class distribution of PerformanceRating in both training and testing datasets, ensuring a fair evaluation.

### Feature Scaling

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Scaling completed successfully.")
````

### Cross-Validation Strategy

````python
from sklearn.model_selection import StratifiedKFold

cv = StratifiedKFold(

    n_splits=5,

    shuffle=True,

    random_state=42

)

print(cv)
````

### Create Model Dictionary

````python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier
)
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBClassifier

models = {

    "Logistic Regression":
        LogisticRegression(max_iter=1000),

    "Decision Tree":
        DecisionTreeClassifier(random_state=42),

    "Random Forest":
        RandomForestClassifier(random_state=42),

    "Gradient Boosting":
        GradientBoostingClassifier(random_state=42),

    "AdaBoost":
        AdaBoostClassifier(random_state=42),

    "XGBoost":
        XGBClassifier(
            objective="multi:softprob",
            random_state=42,
            eval_metric="mlogloss"
        ),

    "Support Vector Machine":
        SVC(probability=True),

    "K Nearest Neighbor":
        KNeighborsClassifier(),

    "Naive Bayes":
        GaussianNB()

}
````

### Train All Models

````python
results = []

for name, model in models.items():

    if name in [
        "Logistic Regression",
        "Support Vector Machine",
        "K Nearest Neighbor"
    ]:

        model.fit(X_train_scaled, y_train)

        predictions = model.predict(X_test_scaled)

    else:

        model.fit(X_train, y_train)

        predictions = model.predict(X_test)

    accuracy = accuracy_score(y_test, predictions)

    precision = precision_score(
        y_test,
        predictions,
        average="weighted"
    )

    recall = recall_score(
        y_test,
        predictions,
        average="weighted"
    )

    f1 = f1_score(
        y_test,
        predictions,
        average="weighted"
    )

    cv_score = cross_val_score(
        model,
        X_train_scaled if name in [
            "Logistic Regression",
            "Support Vector Machine",
            "K Nearest Neighbor"
        ] else X_train,
        y_train,
        cv=cv,
        scoring="accuracy"
    ).mean()

    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1,
        cv_score
    ])
````

### Model Comparison Table

````python
results_df = pd.DataFrame(

    results,

    columns=[

        "Model",

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score",

        "Cross Validation"

    ]

)

results_df.sort_values(

    by="Accuracy",

    ascending=False

)
````

### Reusable Visualization Functions

````python
def plot_count(data, column, title, figsize=(8, 5)):
    """
    Creates a count plot for a categorical variable.
    """
    plt.figure(figsize=figsize)
    sns.countplot(data=data, x=column)
    plt.title(title)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


def plot_box(data, x, y, title, figsize=(8, 5)):
    """
    Creates a boxplot for numerical analysis.
    """
    plt.figure(figsize=figsize)
    sns.boxplot(data=data, x=x, y=y)
    plt.title(title)
    plt.tight_layout()
    plt.show()


def plot_histogram(data, column, bins=20, figsize=(8, 5)):
    """
    Creates a histogram with KDE.
    """
    plt.figure(figsize=figsize)
    sns.histplot(data[column], bins=bins, kde=True)
    plt.title(f"Distribution of {column}")
    plt.tight_layout()
    plt.show()
````
